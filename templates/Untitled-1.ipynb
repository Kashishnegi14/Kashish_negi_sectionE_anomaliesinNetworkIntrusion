```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Simulate traffic data
dates = pd.date_range(start="2025-04-01", periods=30)
traffic = np.random.poisson(lam=500, size=30)

plt.figure(figsize=(10, 5))
plt.plot(dates, traffic, marker='o', linestyle='-', color='blue')
plt.title("Daily Network Traffic Volume")
plt.xlabel("Date")
plt.ylabel("Packets per Day")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
protocols = ['TCP', 'UDP', 'ICMP', 'HTTP', 'HTTPS', 'Other']
usage = [35, 25, 10, 15, 10, 5]

plt.figure(figsize=(7, 7))
plt.pie(usage, labels=protocols, autopct='%1.1f%%', startangle=140)
plt.title("Protocol-wise Traffic Distribution")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
days = np.arange(1, 31)
normal_traffic = np.random.normal(loc=400, scale=50, size=30)
anomalies = np.random.choice([0, 1], size=30, p=[0.8, 0.2]) * np.random.randint(600, 900, size=30)

combined = normal_traffic + anomalies

plt.figure(figsize=(10, 5))
plt.plot(days, combined, label='Total Traffic', color='darkorange')
plt.scatter(days[anomalies > 0], combined[anomalies > 0], color='red', label='Anomaly Detected')
plt.title("Anomaly Detection in Network Traffic")
plt.xlabel("Day")
plt.ylabel("Packets per Day")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

ips = [f"192.168.1.{i}" for i in range(101, 106)]
alerts = np.random.randint(20, 100, size=5)

plt.figure(figsize=(8, 5))
sns.barplot(x=alerts, y=ips, palette='magma')
plt.title("Top 5 Suspicious IPs by Alert Count")
plt.xlabel("Alert Count")
plt.ylabel("IP Address")
plt.show()
```
